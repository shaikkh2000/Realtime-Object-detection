```python
import cv2
import numpy as np

# Load the pre-trained model and class labels
model = cv2.dnn_DetectionModel('frozen_inference_graph.pb', 'ssd_mobilenet_v3_large_coco_2020_01_14.pbtxt')
classLabels = ['person', 'car', 'bicycle', 'dog', 'cat', 'truck', ...]

# Use DroidCam as video source
cap = cv2.VideoCapture('traffic.mp4')  # Try 1, 2, etc., if 0 doesn't work
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)
cap.set(cv2.CAP_PROP_FOURCC, cv2.VideoWriter_fourcc(*'MJPG'))

if not cap.isOpened():
    raise IOError("Cannot open video source")

# Configure the detection model
model.setInputSize(320, 320)
model.setInputScale(1.0 / 127.5)
model.setInputMean((127.5, 127.5, 127.5))
model.setInputSwapRB(True)

font_scale = 1
font = cv2.FONT_HERSHEY_PLAIN
CONFIDENCE_THRESHOLD = 0.6

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to read frame")
        break

    try:
        # Perform object detection
        results = model.detect(frame, confThreshold=CONFIDENCE_THRESHOLD)
        if len(results) == 3:
            ClassIndex, confidence, bbox = results
            for classInd, conf, boxes in zip(ClassIndex, confidence, bbox):
                if classInd <= len(classLabels):
                    cv2.rectangle(frame, boxes, color=(255, 0, 0), thickness=2)
                    label = f"{classLabels[classInd - 1]}: {conf:.2f}"
                    cv2.putText(frame, label, (boxes[0] + 10, boxes[1] + 40),
                                font, fontScale=font_scale, color=(0, 255, 0), thickness=2)
    except Exception as e:
        print(f"Error during detection: {e}")

    cv2.imshow("Object Detection", frame)
    if cv2.waitKey(20) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

```
